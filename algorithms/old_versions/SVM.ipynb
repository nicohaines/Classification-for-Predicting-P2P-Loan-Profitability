```python
# imports
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.svm import SVC
from sklearn.svm import LinearSVC 
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import train_test_split
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                             f1_score, roc_auc_score, classification_report, confusion_matrix)
```

```python
df = pd.read_csv("../dataset/clean_undersampled.csv", low_memory=False)
```

```python
# features / target
y = df['profitable']
X = df.drop(columns=['profitable'])
print(X.shape)
print("class balance:\n", y.value_counts())
```

```python
# same split as XGBoost notebook
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.15, random_state=42, stratify=y)

# validation split off the training portion
X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.1765, random_state=42, stratify=y_train)

print(len(X_train), len(X_val), len(X_test))
print("class balance per split:", y_train.mean(), y_val.mean(), y_test.mean())  # all ~0.5
```

```python
# scale each feature so that large values, like loan amount dont expload in distance
# distance = sqrt(Δgrade² + Δloan_amnt² + Δannual_inc² + ...)
# SVM in one object so the scaler only fits on training data

# SVM training do not take NaN vlaues. SimpleImputer i used to fill those columns

svm_pipeline = make_pipeline(
    
    SimpleImputer(strategy='median'),  # robust to the skewed money columns
    StandardScaler(),      # mean=0, std=1 for every feature
    LinearSVC()  # linear SVM, default C=1
)

# fit ONLY on the training portion
svm_pipeline.fit(X_train, y_train)
#print("fitted. support vectors:", svm_pipeline.named_steps['svm'].n_support_)
```

```python
y_pred = svm_pipeline.predict(X_val)

print("Accuracy:", accuracy_score(y_val, y_pred))
print("Precision:", precision_score(y_val, y_pred))
print("Recall:   ", recall_score(y_val, y_pred))
print("F1:       ", f1_score(y_val, y_pred))
print("ROC AUC:  ", roc_auc_score(y_val, svm_pipeline.decision_function(X_val)))
print()
print(classification_report(y_val, y_pred))
print(confusion_matrix(y_val, y_pred))
```

```python
# --- SVM decision-boundary visualization (2 most important features) ---
# grade (0-6 letter grade) x term_months (36/60). This is a standalone 2D
# SVM used ONLY for illustration -- it is not the full model in svm_pipeline.

feat1, feat2 = "grade", "term_months"

df2 = X_train[[feat1, feat2]]

# standalone 2D pipeline: same scaling + linear SVM as the main model
pipe2 = make_pipeline(StandardScaler(), SVC(kernel="linear", C=1))
pipe2.fit(df2, y_train)

svc = pipe2.named_steps["svc"]
scaler = pipe2.named_steps["standardscaler"]

# mesh grid over a margin beyond the data range
x1 = np.linspace(df2[feat1].min() - 0.5, df2[feat1].max() + 0.5, 400)
x2 = np.linspace(df2[feat2].min() - 15, df2[feat2].max() + 15, 400)
XX, YY = np.meshgrid(x1, x2)

# predict the decision function on the grid using the SAME fitted scaler
grid_pts = np.column_stack([XX.ravel(), YY.ravel()])
Z = svc.decision_function(scaler.transform(grid_pts)).reshape(XX.shape)

fig, ax = plt.subplots(figsize=(8, 6), dpi=150)

# class-colored regions
ax.contourf(XX, YY, Z, levels=[-1, 0, 1], alpha=0.3,
            cmap="coolwarm")

# the hyperplane (zero contour)
ax.contour(XX, YY, Z, levels=[0], colors="black", linewidths=2)

# margin boundaries (+/- 1 on the scaled decision function)
ax.contour(XX, YY, Z, levels=[-1, 1], linestyles="--", colors="gray")

# data points, colored by class
ax.scatter(df2[feat1], df2[feat2], c=y_train.values, cmap="coolwarm",
           s=12, alpha=0.5, edgecolors="none")

# support vectors (transformed back from scaled space)
sv = scaler.inverse_transform(svc.support_vectors_)
ax.scatter(sv[:, 0], sv[:, 1], s=70, facecolors="none",
           edgecolors="black", linewidths=1, label="support vectors")

ax.set_xlabel(feat1)
ax.set_ylabel(feat2)
ax.set_title("Linear SVM decision boundary: grade vs term length")
ax.legend(loc="upper left")
plt.tight_layout()
plt.show()
```
